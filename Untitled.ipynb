```python
spotify_train = pd.read_csv("data/processed/spotify_train.csv")
spotify_train.head()
```

```python
spotify_test = pd.read_csv("data/processed/spotify_test.csv")
spotify_test.head()
```

```python
target = "playlist_genre"

features = [
    "danceability",
    "energy",
    "key",
    "loudness",
    "mode",
    "speechiness",
    "acousticness",
    "instrumentalness",
    "liveness",
    "valence",
    "tempo",
    "duration_ms"
]
```

```python
X_train = spotify_train[features]
y_train = spotify_train[target]

X_test = spotify_test[features]
y_test = spotify_test[target]
```

## Feature Selection

```python
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import mutual_info_classif

# Standardize training features for feature ranking
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)

# Mutual Information scores
mi_scores = mutual_info_classif(
    X_train_scaled,
    y_train,
    discrete_features=[
        False, False, True, False, True, False,
        False, False, False, False, False, False
    ],
    random_state=42
)

# Rank features
feature_ranking = pd.DataFrame({
    "feature": features,
    "mi_score": mi_scores
}).sort_values(
    "mi_score", ascending=False
).reset_index(drop=True)

feature_ranking
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import StratifiedKFold, cross_val_score
```

```python
ranked_features = feature_ranking["feature"].tolist()

feature_subsets = {
    "Top 3": ranked_features[:3],
    "Top 5": ranked_features[:5],
    "Top 8": ranked_features[:8],
    "All 12": ranked_features
}
```

```python
# 5-fold cross-validation
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=522
)

# Candidate values for k
k_values = range(3, 201, 5)

cv_results = []

for subset_name, subset_features in feature_subsets.items():

    for k in k_values:

        model = make_pipeline(
            StandardScaler(),
            KNeighborsClassifier(n_neighbors=k)
        )

        scores = cross_val_score(
            model,
            X_train[subset_features],
            y_train,
            cv=cv,
            scoring="accuracy",
            n_jobs=-1
        )

        cv_results.append({
            "feature_subset": subset_name,
            "n_features": len(subset_features),
            "k": k,
            "mean_cv_accuracy": scores.mean(),
            "std_cv_accuracy": scores.std()
        })

cv_results_df = pd.DataFrame(cv_results)

cv_results_df.head()
```

```python
# Sort by mean CV accuracy
cv_results_df = cv_results_df.sort_values(
    "mean_cv_accuracy",
    ascending=False
).reset_index(drop=True)

# Best configuration
cv_results_df.head(10)
```

```python
best_result = cv_results_df.loc[
    cv_results_df["mean_cv_accuracy"].idxmax()
]

best_result
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

for subset_name in feature_subsets:
    subset_results = cv_results_df[
        cv_results_df["feature_subset"] == subset_name
    ].sort_values("k")

    plt.plot(
        subset_results["k"],
        subset_results["mean_cv_accuracy"],
        marker="o",
        label=subset_name
    )

plt.xlabel("Number of Neighbors (k)")
plt.ylabel("Mean 5-Fold CV Accuracy")
plt.title("KNN Performance Across Feature Subsets and k Values")
plt.legend(title="Feature Subset")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
heatmap_data = cv_results_df.pivot(
    index="feature_subset",
    columns="k",
    values="mean_cv_accuracy"
)

heatmap_data = heatmap_data.reindex(
    ["Top 3", "Top 5", "Top 8", "All 12"]
)

plt.figure(figsize=(30, 10))

plt.imshow(heatmap_data, aspect="auto")

plt.xticks(range(len(heatmap_data.columns)), heatmap_data.columns)
plt.yticks(range(len(heatmap_data.index)), heatmap_data.index)

plt.xlabel("Number of Neighbors (k)")
plt.ylabel("Feature Subset")
plt.title("KNN Cross-Validation Accuracy")

for i in range(len(heatmap_data.index)):
    for j in range(len(heatmap_data.columns)):
        plt.text(
            j, i,
            f"{heatmap_data.iloc[i, j]:.3f}",
            ha="center",
            va="center"
        )

plt.colorbar(label="Mean CV Accuracy")
plt.tight_layout()
plt.show()
```
